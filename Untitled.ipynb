```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = np.array([0, 1, 1, 3, 9, 2, 1, 5, 4, 9, 7])
new_img = np.zeros(image.shape)
image_padded = np.pad(image, 1)

print(image_padded)

kernel = np.array([1/5,1/5,1/5, 1/5, 1/5])

views = np.lib.stride_tricks.sliding_window_view(image_padded, kernel.shape)
for i ,view in enumerate(views):
    print(view)
    new_img[i] = np.sum(view * kernel)
print (new_img)

    
```

```python
plt.imshow(np.expand_dims(new_img, 1) ,cmap = "gray")
```

```python
img = np.arange(25).reshape((5,5))
print(img)
kerneal = np.array([[0, 1/8, 0,],
                  [1/8, 4/8, 1/8 ,],
                  [0, 1/8, 0 ,]])
print (kerneal)
padded_img = np.pad(img, 1)

new_img = np.zeros_like(img)

views = np.lib.stride_tricks.sliding_window_view(padded_img, kerneal.shape)
for i, views_rows in enumerate(views):
    for j, view in enumerate(views_rows):
        new_img[i, j] = np.sum(views[j] * kerneal)
print (new_img)
f, (ax1, ax2) = plt.subplots(1, 2)
ax1.imshow(new_img, cmap = "grey")
ax2.imshow(img, cmap = "grey")
plt.show()
    
    
```
