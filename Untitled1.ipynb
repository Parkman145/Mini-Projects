```python
layer = tf.keras.layers.Conv2D(1, (3, 3), activation='relu')
layer.build(input_shape=(None, 10, 10, 1))
weights = layer.get_weights()
kernel_shape = weights[0].shape
bias_shape = weights[1].shape

layer.set_weights([np.ones(kernel_shape), np.zeros(bias_shape)] )
```

```python
images = np.zeros((1, 10, 10, 1))
images[0, 5, 5, 0] = 10
images_blurred = layer(images)
plt.imshow(images_blurred[0])
print(images_blurred[0])
```
